### 3.1 `crime.mv_d_uzemi`

```python
%%sql
CREATE OR REPLACE MATERIALIZED LAKE VIEW crime.mv_d_uzemi
as
SELECT
obec_text as `název_obce`,
obec_kod as `kód_obce`,
obec_typ as `typ_obce`,
pou_text as `název_POU`,
pou_csu_cis61_kod as `kód_POU_ČSÚ`,
pou_ruian_kod as `kód_POU_RÚIAN`,
pou_sidlo_obec_kod as `kód_obce_sidlo_POU`,
orp_text as `název_ORP`,
orp_csu_cis65_kod as `kód_ORP_ČSÚ`,
orp_ruian_kod as `kód_ORP_RÚIAN`,
orp_sidlo_obec_kod as `kód_obce_sidlo_ORP`,
okres_text as `název_okresu`,
okres_csu_cis101_lau_kod as `kód_okresu_LAU_ČSÚ`,
okres_csu_cis109_nuts_kod as `kód_okresu_NUTS_ČSÚ`,
okres_ruian_kod as `kód_okresu_RÚIAN`,
kraj_text as `název_kraje`,
kraj_zkratka as `zkratka_kraje`,
kraj_csu_cis100_kod as `kód_kraje_ČSÚ`,
kraj_csu_cis108_nuts_kod as `kód_kraje_NUTS_ČSÚ`,
kraj_ruian_vusc_kod as `kód_kraje_VÚSC_RÚIAN`,
region_text as `název_regionu_soudržnosti`,
region_csu_cis99_kod as `kód_regionu_soudržnosti_ČSÚ`,
region_csu_cis107_nuts_kod as `kód_regionu_soudržnosti_NUTS_ČSÚ`,
region_ruian_kod as `kód_regionu_soudržnosti_RÚIAN`,
    regexp_replace(kraj_text, '(?i)(^|\\s)Kraj\\s*', '') AS `název_kraje_short`,
    -- GeoJSON shapes for the report's map visuals (served from the GitHub repo)
    concat('https://raw.githubusercontent.com/Thepilli/czech_crime/refs/heads/main/geodata/split/kod_kraj_csu_', CAST(kraj_csu_cis100_kod AS STRING), '.json') AS `kraj_json_url`,
    concat('https://raw.githubusercontent.com/Thepilli/czech_crime/refs/heads/main/geodata/split_okres/kod_okresu_lau_csu_', CAST(okres_csu_cis101_lau_kod AS STRING), '.json') AS `okres_json_url`
FROM crime.d_uzemi
```

### 3.2 `crime.mv_f_crime`

```python
%%sql
CREATE OR REPLACE MATERIALIZED LAKE VIEW crime.mv_f_crime
AS
SELECT
    crime_date AS CrimeDate,   -- Europe/Prague local date
    time_key   AS TimeKey,     -- minute of day 0-1439 -> d_time[TimeKey]
    mp,
    state,
    relevance,
    kod_zuj,
    kod_obec,
    kod_momc,
    type_group_key,
    primary_type
FROM crime.f_crime
```

### 3.3 `crime.mv_d_date`

```python
%%sql
CREATE OR REPLACE MATERIALIZED LAKE VIEW crime.mv_d_date AS
SELECT
  Date,
  Year,
  YearQuarter,
  Quarter,
  yearmonth,
  YearMonthNumber,
  Month,
  MonthName,
  MonthNameCs,
  MonthLetter,
  MonthLetterCs,
  DayofMonth,
  DayofWeek,
  DayName,
  DayNameCs,
  isWeekend
FROM crime.d_date;
```
